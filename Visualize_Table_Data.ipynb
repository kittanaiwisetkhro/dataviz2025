```python
from colorspacious import cspace_converter

import matplotlib.pyplot as plt
import numpy as np

import matplotlib as mpl
```

```python
def plot_color_gradients(category, cmap_list):
    # Create figure and axes
    fig, axes = plt.subplots(nrows=len(cmap_list), figsize=(4, len(cmap_list) * 0.5))
    fig.suptitle(category, fontsize=14, y=1.02) # Add category as super title

    # Plot each colormap
    for ax, cmap in zip(axes, cmap_list):
        gradient = np.linspace(0, 1, 256)
        gradient = np.vstack((gradient, gradient))
        ax.imshow(gradient, aspect='auto', cmap=mpl.colormaps[cmap])
        ax.set_title(cmap, fontsize=10, loc='right') # Add colormap name as title
        ax.set_axis_off() # Turn off axes

    plt.tight_layout() # Adjust layout
    plt.show() # Display the figure

plot_color_gradients('Perceptually Uniform Sequential',
                     ['viridis', 'plasma', 'inferno', 'magma', 'cividis'])
```
