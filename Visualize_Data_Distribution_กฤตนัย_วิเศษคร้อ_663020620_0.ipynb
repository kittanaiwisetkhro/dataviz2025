````
stopandsearch_df_dropna = stopandsearch_df.dropna(subset= ['Age range','Gender' ])
````

````
# replace the values in 'Gender' column with 0 for Femali and 1 For Male

stopandsearch_df_dropna['Gender'] = stopandsearch_df_dropna['Gender'].replace({
    'Female': 1,
    'Male': 35})
stopandsearch_df_dropna.head()
````

````
# replace the values in 'Age range'' column with 14 for 10-17 , 21 For 18-24 , 30 For 25-34 , 34 For over 34 , 10 For under 10

stopandsearch_df_dropna['Age range'] = stopandsearch_df_dropna['Age range'].replace({
    '10-17': 14,
    '18-24': 21,
    '25-34': 30,
    'over 34': 34,
    'under 10': 10
})
stopandsearch_df_dropna.head()
````

plot กราฟ 3 อันในแนวแทยงมุม
1. boxplot

3. histogram

````
fig, axes = plt.subplots(3, 3, figsize=(15, 15))

# Boxplot (top-left)
axes[0, 0].boxplot(stopandsearch_df['Latitude'].dropna())
axes[0, 0].set_title('Latitude Boxplot')
axes[0, 0].set_ylabel('Latitude')

# Scatterplot (middle) - using Latitude for both axes to show distribution
axes[1, 1].scatter(stopandsearch_df['Latitude'].dropna(), stopandsearch_df['Latitude'].dropna(), alpha=0.3)
axes[1, 1].set_title('Latitude Scatterplot')
axes[1, 1].set_xlabel('Latitude')
axes[1, 1].set_ylabel('Latitude')

# Histogram (bottom-right)
axes[2, 2].hist(stopandsearch_df['Latitude'].dropna())
axes[2, 2].set_title('Latitude Histogram')
axes[2, 2].set_xlabel('Latitude')
axes[2, 2].set_ylabel('Frequency')

# Hide the other subplots
for i in range(3):
    for j in range(3):
        if i != j:
            axes[i, j].axis('off')

plt.tight_layout()
plt.show()
````

````
fig, axes = plt.subplots(3, 3, figsize=(15, 15))

# Boxplot (top-left)
axes[0, 0].boxplot(stopandsearch_df['Latitude'].dropna())
axes[0, 0].set_title('Latitude Boxplot')

# Scatterplot (middle)
axes[1, 1].scatter(stopandsearch_df_dropna['Age range'], stopandsearch_df_dropna['Gender'], alpha=0.3)
axes[1, 1].set_title('Age vs Gender Scatterplot')
axes[1, 1].set_xlabel('Age range')
axes[1, 1].set_ylabel('Gender')

# Histogram (bottom-right)
axes[2, 2].hist(stopandsearch_df['Latitude'].dropna())
axes[2, 2].set_title('Latitude Histogram')
axes[2, 2].set_xlabel('Latitude')

# Hide the other subplots
for i in range(3):
    for j in range(3):
        if i != j:
            axes[i, j].axis('off')

plt.tight_layout()
plt.show()
````
