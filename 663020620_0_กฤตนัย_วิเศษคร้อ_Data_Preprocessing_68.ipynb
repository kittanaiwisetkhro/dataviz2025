### จัดการ Missing values

* (10 points) จากข้อมูลที่ให้มา เตรียมข้อมูลให้พร้อมโดยการจัดการกับค่าว่างในข้อมูล ให้อธิบายวิธีที่ท่านเลือกใช้ พร้อมยกเหตุผลประกอบ

**จัดการ Missing values**

จากข้อมูลที่ให้มา ได้ทำการจัดการกับค่าว่างในข้อมูลดังนี้:

*   **Age:** ค่าว่างในคอลัมน์ 'age' ถูกแทนที่ด้วยค่ามัธยฐาน (median) ของคอลัมน์ เนื่องจากอายุเป็นข้อมูลเชิงตัวเลขและค่ามัธยฐานมีความทนทานต่อค่าผิดปกติ (outliers) มากกว่าค่าเฉลี่ย
*   **Fare:** ค่าว่างในคอลัมน์ 'fare' ถูกแทนที่ด้วยค่ามัธยฐาน (median) เช่นเดียวกับ 'age' ด้วยเหตุผลเดียวกัน
*   **Embarked:** ค่าว่างในคอลัมน์ 'embarked' (ท่าเรือที่ขึ้นเรือ) ถูกแทนที่ด้วยค่าฐานนิยม (mode) ซึ่งเป็นค่าที่ปรากฏบ่อยที่สุด เนื่องจาก 'embarked' เป็นข้อมูลเชิงคุณภาพ การแทนที่ด้วยค่าที่พบบ่อยที่สุดเป็นวิธีที่เหมาะสม
*   **Cabin, Boat, Home.dest:** คอลัมน์เหล่านี้มีจำนวนค่าว่างเป็นจำนวนมาก:
    *   'cabin' ถูกแทนที่ด้วยค่า 'Unknown' เพื่อระบุว่าข้อมูลห้องโดยสารไม่ทราบหรือไม่พร้อมใช้งาน การแทนที่ด้วยค่าคงที่เป็นวิธีหนึ่งในการจัดการกับค่าว่างจำนวนมากโดยไม่ต้องลบคอลัมน์ทั้งหมด
    *   'boat' และ 'home.dest' ก็ถูกแทนที่ด้วยค่า 'Unknown' ด้วยเหตุผลคล้ายคลึงกับ 'cabin' อย่างไรก็ตาม ควรพิจารณาว่าคอลัมน์เหล่านี้อาจมีประโยชน์น้อยในการวิเคราะห์เนื่องจากมีค่าว่างจำนวนมาก และอาจพิจารณาทางเลือกอื่นเช่นการลบคอลัมน์หากไม่สามารถใช้ข้อมูลที่มีอยู่ได้อย่างมีประสิทธิภาพ
    *   'body' มีจำนวนค่าว่างสูงที่สุดและอาจไม่มีความเกี่ยวข้องโดยตรงกับการรอดชีวิต จึงตัดสินใจลบคอลัมน์นี้ออกจาก DataFrame

```python
# Fill missing 'age' values with the median age
titanic_df['age'].fillna(titanic_df['age'].median(), inplace=True)

# Fill missing 'fare' values with the median fare
titanic_df['fare'].fillna(titanic_df['fare'].median(), inplace=True)

# Fill missing 'embarked' values with the mode (most frequent value)
titanic_df['embarked'].fillna(titanic_df['embarked'].mode()[0], inplace=True)

# For columns with a large number of missing values like 'cabin', 'boat', 'body', and 'home.dest',
# we might consider dropping them or using a different strategy.
# For now, let's fill 'cabin' with a placeholder indicating missing information.
titanic_df['cabin'].fillna('Unknown', inplace=True)

# You might want to consider dropping 'body', 'boat', and 'home.dest' due to the high number of missing values,
# or use a more sophisticated imputation method if these columns are deemed important.
# For this example, let's fill 'boat' and 'home.dest' with placeholders and drop 'body'.

titanic_df['boat'].fillna('Unknown', inplace=True)
titanic_df['home.dest'].fillna('Unknown', inplace=True)
titanic_df.drop('body', axis=1, inplace=True)

```

```python
print("Missing values after handling:")
display(titanic_df.isnull().sum())
```

## แปลงข้อมูลเชิงคุณภาพให้เป็นตัวเลข

* แปลงข้อมูลเชิงคุณภาพให้เป็นตัวเลข \

*   รายการในลิสต์
*   รายการในลิสต์


    i. (10 points) ใช้ one-hot encoding ในการแปลงตัวแปร Embarked \
    ii. (10 points) แปลงตัวแปร Sex โดยให้เพศชายเป็น 0 และเพศหญิงเป็น 1 \
    iii. (10 points) สร้างตัวแปร FamilySize ซึ่งเป็นผลรวมของตัวแปร SibSp (number of siblings/spouses aboard) และ Parch (number of parents/children aboard) \
    iv. (10 points) สร้างตัวแปร IsAlone ซึ่งมีค่าเป็น 1 ถ้า FamilySize เป็น 0, และมีค่าเป็นศูนย์เมื่อเป็นอื่น

```python
# 1. แปลงข้อมูลเชิงคุณภาพให้เป็นตัวเลข ใช้ one-hot encoding ในการแปลงตัวแปร Embarked

# Apply one-hot encoding to the 'embarked' column
embarked_one_hot = pd.get_dummies(titanic_df['embarked'], prefix='embarked')

# Concatenate the new one-hot encoded columns to the original DataFrame
titanic_df = pd.concat([titanic_df, embarked_one_hot], axis=1)

# Drop the original 'embarked' column as it's no longer needed
titanic_df.drop('embarked', axis=1, inplace=True)

print("DataFrame after one-hot encoding 'embarked':")
display(titanic_df.head())
```

```python
# Convert 'sex' column to numerical: male=0, female=1
titanic_df['sex'] = titanic_df['sex'].map({'male': 0, 'female': 1})

print("DataFrame after converting 'sex' to numerical:")
display(titanic_df.head())
```

```python
# สร้างตัวแปร FamilySize ซึ่งเป็นผลรวมของตัวแปร SibSp (number of siblings/spouses aboard) และ Parch (number of parents/children aboard)

# Create 'FamilySize' as the sum of 'SibSp' and 'Parch'
titanic_df['FamilySize'] = titanic_df['sibsp'] + titanic_df['parch']

print("DataFrame with 'FamilySize' column:")
display(titanic_df.head())
```

```python
# Create 'IsAlone' column: 1 if FamilySize is 0, 0 otherwise
titanic_df['IsAlone'] = (titanic_df['FamilySize'] == 0).astype(int)

print("DataFrame with 'IsAlone' column:")
display(titanic_df.head())
```

## ทำการ Normalize ข้อมูล

* ทำการ Normalize ข้อมูล \
    i. (10 points) ตัวแปร Fare และ Age ใช้ Min-Max Scaling \
    ii. (10 points) สร้างตัวแปรใหม่ Fare_Per_Person ด้วยการนำตัวแปร Fare มาหารด้วย FamilySize + 1 \
    iii. (10 points) สร้างตัวแปร AgeGroup ที่ระบุกลุ่มอายุของตัวอย่าง โดยแบ่งเป็นกลุ่มดังนี้ [0, 12), [12, 18), [18, 50) และ [50, 100)

```python
# 1. ทำการ Normalize ข้อมูล ตัวแปร Fare และ Age ใช้ Min-Max Scaling

from sklearn.preprocessing import MinMaxScaler

# Initialize the MinMaxScaler
scaler = MinMaxScaler()

# Apply Min-Max scaling to 'Fare' and 'Age' columns
titanic_df[['fare', 'age']] = scaler.fit_transform(titanic_df[['fare', 'age']])

print("DataFrame after Min-Max scaling 'Fare' and 'Age':")
display(titanic_df.head())
```

```python
# 2. ทำการ Normalize ข้อมูล สร้างตัวแปรใหม่ Fare_Per_Person ด้วยการนำตัวแปร Fare มาหารด้วย FamilySize + 1

# Create 'Fare_Per_Person' by dividing 'Fare' by 'FamilySize + 1'
titanic_df['Fare_Per_Person'] = titanic_df['fare'] / (titanic_df['FamilySize'] + 1)

print("DataFrame with 'Fare_Per_Person' column:")
display(titanic_df.head())
```

```python
# Hint ข้อ iii. สามารถใช้ฟังก์ชัน cut ใน pandas ช่วยได้
import pandas as pd
import numpy as np
pd.cut(np.array([1, 7, 5, 4, 6, 3]), 3) # แบ่งข้อมูลทั้งหมดเป็น 3 ช่วง
# [(0.994, 3.0], (5.0, 7.0], (3.0, 5.0], (3.0, 5.0], (5.0, 7.0], (0.994, 3.0]]
# Categories (3, interval[float64, right]): [(0.994, 3.0] < (3.0, 5.0] < (5.0, 7.0]]
pd.cut(np.array([1, 7, 5, 4, 6, 3]), [0,2,5,8]) # แบ่งข้อมูลทั้งหมดตามค่าที่กำหนด
# [(0, 2], (5, 8], (2, 5], (2, 5], (5, 8], (2, 5]]
# Categories (3, interval[int64, right]): [(0, 2] < (2, 5] < (5, 8]]
```

```python
# 3. ทำการ Normalize ข้อมูล สร้างตัวแปร AgeGroup ที่ระบุกลุ่มอายุของตัวอย่าง โดยแบ่งเป็นกลุ่มดังนี้ [0, 12), [12, 18), [18, 50) และ [50, 100)

# Create 'AgeGroup' by categorizing 'age' into specified bins
bins = [0, 12, 18, 50, 100]
labels = ['0-11', '12-17', '18-49', '50-99']
titanic_df['AgeGroup'] = pd.cut(titanic_df['age'], bins=bins, labels=labels, right=False)

print("DataFrame with 'AgeGroup' column:")
display(titanic_df.head())
```

## นำเสนอการแจกแจง/ทำความเข้าใจตัวแปรสำคัญในข้อมูล

## หาความสัมพันธ์ของตัวแป

* เมื่อเตรียมข้อมูลพร้อมแล้ว ให้หาความสัมพันธ์ของตัวแปรต่อไปนี้จากข้อมูล ซึ่งอาจนำเสนอด้วยสถิติและแผนภูมิหรือกราฟก็ได้ พร้อมอธิบายสั้น ๆ ว่าพบอะไรหรือสามารถสรุปอะไรได้จากภาพที่ได้ \
    i. (10 points) survived และ pclass \
    ii. (10 points) survived และ sex \
    iii. (10 points) survived และ age \
    iv. (10 points) จำนวนสมาชิกครอบครัวที่รอดชีวิต และ Fare

```python
# 1. หาความสัมพันธ์ของตัวแปร survived และ pclass นำเสนอด้วยสถิติและแผนภูมิหรือกราฟ

# Relationship between survived and pclass
print("Survival rate by Pclass:")
display(titanic_df.groupby('pclass')['survived'].mean())

# Visualize the relationship
plt.figure(figsize=(6, 4))
sns.barplot(x='pclass', y='survived', data=titanic_df)
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.show()
```

```python
# 2. หาความสัมพันธ์ของตัวแปร survived และ sex นำเสนอด้วยสถิติและแผนภูมิหรือกราฟ

# Relationship between survived and sex
print("Survival rate by Sex:")
display(titanic_df.groupby('sex')['survived'].mean())

# Visualize the relationship
plt.figure(figsize=(6, 4))
sns.barplot(x='sex', y='survived', data=titanic_df)
plt.title('Survival Rate by Sex (0: Male, 1: Female)')
plt.xlabel('Sex')
plt.ylabel('Survival Rate')
plt.xticks([0, 1], ['Male', 'Female'])
plt.show()
```

```python
# 3. หาความสัมพันธ์ของตัวแปร survived และ age นำเสนอด้วยสถิติและแผนภูมิหรือกราฟ

# Relationship between survived and age

print("Age distribution by survival status:")
display(titanic_df.groupby('survived')['age'].describe())

# Visualize the relationship with a violin plot
plt.figure(figsize=(8, 6))
sns.violinplot(x='survived', y='age', data=titanic_df)
plt.title('Age Distribution by Survival Status (0: Not Survived, 1: Survived)')
plt.xlabel('Survived')
plt.ylabel('Age')
plt.xticks([0, 1], ['Not Survived', 'Survived'])
plt.show()
```

```python
# 3. Relationship between survived and AgeGroup/Pclass using a pivot table

print("Survival rate by AgeGroup and Pclass:")
pivot_table_age_pclass = titanic_df.pivot_table(values='survived', index='AgeGroup', columns='pclass', aggfunc='mean')
display(pivot_table_age_pclass)

# You can also visualize this pivot table using a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(pivot_table_age_pclass, annot=True, fmt=".2f", cmap="YlGnBu")
plt.title('Survival Rate by Age Group and Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Age Group')
plt.show()
```

```python
# 4. Relationship between FamilySize (and survival) and Fare

# Scatter plot of Fare vs FamilySize, colored by survival
plt.figure(figsize=(10, 6))
sns.scatterplot(x='FamilySize', y='fare', hue='survived', data=titanic_df, alpha=0.6)
plt.title('Fare vs FamilySize by Survival Status')
plt.xlabel('Family Size')
plt.ylabel('Fare (Normalized)')
plt.legend(title='Survived', loc='upper right', labels=['No', 'Yes'])
plt.show()

# You could also look at the average fare per person based on survival and family size
print("\nAverage Fare_Per_Person by FamilySize and Survival:")
pivot_table_fare_family_survived = titanic_df.pivot_table(values='Fare_Per_Person', index='FamilySize', columns='survived', aggfunc='mean')
display(pivot_table_fare_family_survived)
```

```python
# 4. Relationship between FamilySize (and survival) and Fare

# Grouped bar plot of average Fare_Per_Person by FamilySize and Survival
plt.figure(figsize=(12, 6))
sns.barplot(x='FamilySize', y='Fare_Per_Person', hue='survived', data=titanic_df)
plt.title('Average Fare Per Person by Family Size and Survival Status')
plt.xlabel('Family Size')
plt.ylabel('Average Fare Per Person (Normalized)')
plt.legend(title='Survived', loc='upper right', labels=['No', 'Yes'])
plt.show()
```

**สรุปผลจากตารางสรุปข้อมูล:**

จากตารางสรุปข้อมูลที่แสดงอัตราการรอดชีวิตตาม Pclass, Sex และ AgeGroup เราสามารถสังเกตได้ดังนี้:

*   **Pclass:** ผู้โดยสารชั้น 1 มีอัตราการรอดชีวิตสูงที่สุด ตามมาด้วยชั้น 2 และชั้น 3 อย่างชัดเจน นี่แสดงให้เห็นว่าชั้นที่นั่งมีผลอย่างมากต่อโอกาสในการรอดชีวิต
*   **Sex:** ผู้หญิงมีอัตราการรอดชีวิตสูงกว่าผู้ชายอย่างมีนัยสำคัญในทุก Pclass และ AgeGroup นี่สอดคล้องกับนโยบาย "ผู้หญิงและเด็กก่อน" ในสถานการณ์ฉุกเฉิน
*   **AgeGroup:** แนวโน้มอัตราการรอดชีวิตแตกต่างกันไปในแต่ละกลุ่มอายุและ Pclass อย่างไรก็ตาม โดยทั่วไปแล้ว เด็กเล็ก (0-11 ปี) มักมีอัตราการรอดชีวิตสูงกว่าในบาง Pclass โดยเฉพาะในชั้น 1 และ 2
*   **Interaction between variables:** การรวมกันของปัจจัยต่างๆ มีผลต่ออัตราการรอดชีวิตอย่างเห็นได้ชัด ตัวอย่างเช่น ผู้หญิงในชั้น 1 มีโอกาสรอดชีวิตเกือบ 100% ในขณะที่ผู้ชายในชั้น 3 มีโอกาสรอดชีวิตต่ำมาก

โดยสรุป ปัจจัยที่มีอิทธิพลอย่างมากต่อการรอดชีวิตคือ Pclass และ Sex โดยผู้โดยสารในชั้นที่สูงกว่าและผู้หญิงมีแนวโน้มที่จะรอดชีวิตมากกว่า
